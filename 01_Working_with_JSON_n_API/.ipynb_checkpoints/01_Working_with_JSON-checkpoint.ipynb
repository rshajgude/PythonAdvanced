```python
json_string='''
{
    "name":"Sachin",
    "age":42,
    "sport":"cricket"
}
'''
json_string
```

```python
type(json_string)
```

```python
import json
```

```python
json_obj = json.loads(json_string)
json_obj
```

```python
type(json_obj)
```

### JSON is very similar to dictionary but in python json is a dict like <b>STRING<b>. 
To convert to/from dict to json we use json.loads and json.dumps functions.

```python
dict_obj={
    'name':'Raahul',
    'age':43,
    'runs':14000,
    'matches':267
}
dict_obj
```

```python
type(dict_obj)
```

```python
json_obj_1=json.dumps(dict_obj)
json_obj_1
```

```python
type(json_obj_1)
```
